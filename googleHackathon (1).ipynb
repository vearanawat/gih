##**COMPARISON**

```python
import time
import cv2
import numpy as np
import pytesseract
import matplotlib.pyplot as plt
from paddleocr import PaddleOCR
import easyocr
from Levenshtein import distance as levenshtein_distance

# Initialize OCR models
paddle_ocr = PaddleOCR(use_angle_cls=True, lang="en")
easy_ocr = easyocr.Reader(["en"])

# Function to calculate CER and WER
def calculate_metrics(gt_text, ocr_text):
    cer = levenshtein_distance(gt_text, ocr_text) / max(len(gt_text), 1)
    wer = levenshtein_distance(gt_text.split(), ocr_text.split()) / max(len(gt_text.split()), 1)
    return round(cer, 4), round(wer, 4)

# Function to run OCR and measure performance
def run_ocr(image_path, ground_truth):
    results = {}

    # Load image (Ensure it's a valid format)
    img = cv2.imread(image_path)

    if img is None:
        raise ValueError("Error: Image not found. Check the image path.")

    # PaddleOCR
    start_time = time.time()
    paddle_result = paddle_ocr.ocr(image_path, cls=True)
    paddle_text = " ".join([word[1][0] for line in paddle_result for word in line])
    paddle_time = time.time() - start_time
    paddle_cer, paddle_wer = calculate_metrics(ground_truth, paddle_text)
    results["PaddleOCR"] = {"CER": paddle_cer, "WER": paddle_wer, "Time": paddle_time}

    # EasyOCR (Fix: Pass NumPy array instead of path)
    start_time = time.time()
    easy_result = easy_ocr.readtext(img, detail=0)
    easy_text = " ".join(easy_result)
    easy_time = time.time() - start_time
    easy_cer, easy_wer = calculate_metrics(ground_truth, easy_text)
    results["EasyOCR"] = {"CER": easy_cer, "WER": easy_wer, "Time": easy_time}



    return results


# Function to plot results
def plot_metrics(results):
    models = list(results.keys())
    cer_values = [results[m]["CER"] for m in models]
    wer_values = [results[m]["WER"] for m in models]
    time_values = [results[m]["Time"] for m in models]

    fig, axes = plt.subplots(1, 3, figsize=(15, 5))

    # Plot CER
    axes[0].bar(models, cer_values, color=['blue', 'green', 'red'])
    axes[0].set_title("Character Error Rate (CER)")
    axes[0].set_ylabel("Error Rate")

    # Plot WER
    axes[1].bar(models, wer_values, color=['blue', 'green', 'red'])
    axes[1].set_title("Word Error Rate (WER)")
    axes[1].set_ylabel("Error Rate")

    # Plot Processing Time
    axes[2].bar(models, time_values, color=['blue', 'green', 'red'])
    axes[2].set_title("Processing Time")
    axes[2].set_ylabel("Time (seconds)")

    plt.tight_layout()
    plt.show()

# Example usage
image_path = "/content/Screenshot 2025-02-25 012649.png"  # Replace with your image path
ground_truth = "Your expected text here"  # Replace with actual ground truth text

results = run_ocr(image_path, ground_truth)

# Show results in text format
for model, metrics in results.items():
    print(f"\n{model}: CER={metrics['CER']}, WER={metrics['WER']}, Time={metrics['Time']} sec")

# Plot comparison
plot_metrics(results)
```

```python
import easyocr
import cv2
import matplotlib.pyplot as plt
import time
import numpy as np

def perform_indian_languages_ocr(image_path):
    """
    Perform OCR on Indian languages by processing with separate language pairs
    to overcome EasyOCR's compatibility limitations.
    """
    # Define language pairs (each with English)
    language_pairs = [
        ['en', 'hi'],  # English + Hindi
        ['en', 'bn'],  # English + Bengali
        ['en', 'ta'],  # English + Tamil
        ['en', 'te'],  # English + Telugu
        ['en', 'kn'],  # English + Kannada
    ]

    all_results = []

    # Process with each language pair
    for lang_pair in language_pairs:
        print(f"Processing with languages: {lang_pair}")
        try:
            start_time = time.time()
            reader = easyocr.Reader(
                lang_pair,
                gpu=True if cv2.cuda.getCudaEnabledDeviceCount() > 0 else False,
                download_enabled=True
            )
            print(f"Model loaded in {time.time() - start_time:.2f} seconds")

            # Perform OCR
            results = reader.readtext(image_path)
            print(f"Detected {len(results)} text regions")
            all_results.extend(results)

        except Exception as e:
            print(f"Error processing {lang_pair}: {e}")

    # Remove duplicates based on bounding box overlap
    final_results = remove_duplicates(all_results)
    print(f"After removing duplicates: {len(final_results)} unique text regions")

    # Load image for visualization
    img = cv2.imread(image_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Extract text and draw bounding boxes
    extracted_text = []
    for (bbox, text, prob) in final_results:
        # Extract text with confidence score
        extracted_text.append((text, prob))

        # Draw bounding box
        top_left = tuple(map(int, bbox[0]))
        bottom_right = tuple(map(int, bbox[2]))
        cv2.rectangle(img, top_left, bottom_right, (0, 255, 0), 2)

        # Add text below the box
        cv2.putText(img, f"{text} ({prob:.2f})",
                   (top_left[0], bottom_right[1] + 20),
                   cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 0, 255), 2)

    return extracted_text, img

def remove_duplicates(results, iou_threshold=0.5):
    """
    Remove duplicate detections based on IoU (Intersection over Union)
    of bounding boxes and confidence scores.
    """
    if not results:
        return []

    # Sort by confidence (highest first)
    sorted_results = sorted(results, key=lambda x: x[2], reverse=True)

    final_results = []
    used_boxes = []

    for result in sorted_results:
        bbox, text, prob = result

        # Convert to [x1, y1, x2, y2] format
        box = [
            min(p[0] for p in bbox), min(p[1] for p in bbox),  # top left
            max(p[0] for p in bbox), max(p[1] for p in bbox)   # bottom right
        ]

        # Check if this box overlaps significantly with any used box
        is_duplicate = False
        for used_box in used_boxes:
            if calculate_iou(box, used_box) > iou_threshold:
                is_duplicate = True
                break

        # If not a duplicate, add to final results
        if not is_duplicate:
            final_results.append(result)
            used_boxes.append(box)

    return final_results

def calculate_iou(box1, box2):
    """Calculate Intersection over Union (IoU) between two boxes."""
    # Extract coordinates
    x1_1, y1_1, x2_1, y2_1 = box1
    x1_2, y1_2, x2_2, y2_2 = box2

    # Calculate intersection area
    x_left = max(x1_1, x1_2)
    y_top = max(y1_1, y1_2)
    x_right = min(x2_1, x2_2)
    y_bottom = min(y2_1, y2_2)

    if x_right < x_left or y_bottom < y_top:
        return 0.0

    intersection_area = (x_right - x_left) * (y_bottom - y_top)

    # Calculate union area
    box1_area = (x2_1 - x1_1) * (y2_1 - y1_1)
    box2_area = (x2_2 - x1_2) * (y2_2 - y1_2)
    union_area = box1_area + box2_area - intersection_area

    return intersection_area / union_area if union_area > 0 else 0.0

def display_results(extracted_text, annotated_img=None):
    """Display OCR results and optionally the annotated image."""
    print("\n=== Extracted Text ===")
    for idx, (text, confidence) in enumerate(extracted_text, 1):
        print(f"{idx}. '{text}' (Confidence: {confidence:.2f})")

    # Display the annotated image if provided
    if annotated_img is not None:
        plt.figure(figsize=(15, 15))
        plt.imshow(annotated_img)
        plt.title("Indian Languages OCR Results")
        plt.axis('off')
        plt.show()

# Example usage
# image_path = "/content/Screenshot 2025-02-25 012649.png"
image_path="/content/Fd6D1jVagAANm4r.jpeg"

# Perform OCR
extracted_text, annotated_img = perform_indian_languages_ocr(image_path)

# Display results with visualization
display_results(extracted_text, annotated_img)

# Get plain text only (if needed)
plain_text = [text for text, _ in extracted_text]
print("\n=== Plain Text Only ===")
print("\n".join(plain_text))
```

```python
import numpy as np
import cv2
import torch
from transformers import AutoModelForImageClassification, AutoProcessor, pipeline
from PIL import Image
from huggingface_hub import login
import logging

# Set up logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Log in to Hugging Face
login(token="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")

# Dictionary of fallback models for each type
MODEL_OPTIONS = {
    "xray": [
        "nateraw/xray-diagnosis",  # Original model (not found)
        "medicalai/clinical-xray-diagnosis",  # Alternative medical X-ray model
        "microsoft/resnet-50"  # General purpose fallback
    ],
    "ct": [
        "nateraw/ct-scan-diagnosis",  # Original model (not found)
        "microsoft/swin-tiny-patch4-window7-224",  # General purpose alternative
        "google/vit-base-patch16-224"  # Another general purpose fallback
    ]
}

# Load models with fallback options
def load_model_with_fallback(model_type):
    model_options = MODEL_OPTIONS.get(model_type, [])
    last_exception = None

    for model_name in model_options:
        try:
            logger.info(f"Attempting to load {model_type} model: {model_name}")
            model = AutoModelForImageClassification.from_pretrained(model_name)
            processor = AutoProcessor.from_pretrained(model_name)
            logger.info(f"Successfully loaded {model_name}")
            return model, processor
        except Exception as e:
            logger.warning(f"Failed to load {model_name}: {e}")
            last_exception = e

    # If we've exhausted all options
    raise RuntimeError(f"Failed to load any {model_type} model. Last error: {last_exception}")

# Load models
try:
    xray_model, xray_processor = load_model_with_fallback("xray")
    ct_model, ct_processor = load_model_with_fallback("ct")
    disease_model = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")
    logger.info("All models loaded successfully!")
except Exception as e:
    logger.error(f"Error loading models: {e}")
    raise

# Function to predict X-ray result
def predict_xray(image_path):
    try:
        image = Image.open(image_path).convert("RGB")
        inputs = xray_processor(images=image, return_tensors="pt")

        with torch.no_grad():
            outputs = xray_model(**inputs)

        logits = outputs.logits[0]
        predicted_class = torch.argmax(logits).item()

        return {
            "prediction": xray_model.config.id2label[predicted_class],
        }
    except Exception as e:
        logger.error(f"Error predicting X-ray: {e}")
        return {"error": str(e)}

# Function to predict CT scan result
def predict_ct(image_path):
    try:
        image = Image.open(image_path).convert("RGB")
        inputs = ct_processor(images=image, return_tensors="pt")

        with torch.no_grad():
            outputs = ct_model(**inputs)

        logits = outputs.logits[0]
        predicted_class = torch.argmax(logits).item()

        return {
            "prediction": ct_model.config.id2label[predicted_class],
        }
    except Exception as e:
        logger.error(f"Error predicting CT scan: {e}")
        return {"error": str(e)}

# Function to predict disease based on symptoms
def predict_disease(symptoms):
    try:
        candidate_labels = ["COVID-19", "Flu", "Common Cold", "Pneumonia", "Bronchitis"]
        result = disease_model(symptoms, candidate_labels)
        return {"disease_prediction": result['labels'][0]}
    except Exception as e:
        logger.error(f"Error predicting disease: {e}")
        return {"error": str(e)}

# Test the functions
if __name__ == "__main__":
    xray_image_path = "/content/1-s2.0-S1684118220300682-main.pdf-002-a1.png"
    ct_image_path = "/content/1-s2.0-S1684118220300682-main.pdf-002-a1.png"

    # ct_image_path = "test_ct.jpg"

    print("X-ray Prediction:", predict_xray(xray_image_path))
    print("CT Scan Prediction:", predict_ct(ct_image_path))

    symptoms = "I have a fever, cough, and difficulty breathing."
    print("Disease Prediction:", predict_disease(symptoms))
```
